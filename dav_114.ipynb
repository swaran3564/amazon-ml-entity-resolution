```python

import pandas as pd
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler

# 1. Data Collection
data = load_iris()
df = pd.DataFrame(data.data, columns=data.feature_names)
df["Species"] = data.target_names[data.target]

# 2. Data Cleaning
df.dropna(inplace=True)

# 3. Data Transformation
df["Petal Area"] = df["petal length (cm)"] * df["petal width (cm)"]

# 4. Normalization
cols = data.feature_names
norm = df.copy()
norm[cols] = MinMaxScaler().fit_transform(df[cols])

# 5. Data Interpretation
print("Dataset Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("\nPetal Area:", df["Petal Area"].head(3).tolist())
print("\nNormalized Range:",
      norm[cols].min().min(), "-", norm[cols].max().max())
print("\nAverage Petal Length:")
print(df.groupby("Species")["petal length (cm)"].mean().round(2))
```

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset
df = sns.load_dataset("iris")

# Create Visualizations
fig, ax = plt.subplots(2, 2, figsize=(10, 7))

# 1. Histogram
ax[0,0].hist(df["sepal_length"], bins=10)
ax[0,0].set_title("Histogram")

# 2. Bar Chart
df["species"].value_counts().plot(kind="bar", ax=ax[0,1])
ax[0,1].set_title("Species Count")

# 3. Scatter Plot
sns.scatterplot(data=df, x="sepal_length",
                y="petal_length", hue="species", ax=ax[1,0])
ax[1,0].set_title("Scatter Plot")

# 4. Box Plot
sns.boxplot(data=df, x="species",
            y="petal_length", ax=ax[1,1])
ax[1,1].set_title("Box Plot")

plt.tight_layout()
plt.show()
```
